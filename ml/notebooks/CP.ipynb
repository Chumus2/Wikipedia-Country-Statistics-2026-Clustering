```python
ORIGINAL_DF_PATH = "../data/raw/raw-data.csv"

original_df = load_csv_df(
    df_path=ORIGINAL_DF_PATH
)

df.index = original_df["Country name"]
cluster_profiles = df.groupby("Cluster").mean(numeric_only=True)
display(cluster_profiles.head(10))
```

```python
for cluster_id in sorted(df["Cluster"].unique()):
    countries = df[df["Cluster"] == cluster_id].index.tolist()

    if cluster_id == -1:
        print(f"Noise: {len(countries)} countries")
    else:
        print(f"Cluster {cluster_id}: {len(countries)} countries")
    print(", ".join(map(str, countries)))
```

```python
OUTPUT_PATH = "../data/final/cluster_profiles.csv"

cluster_profiles.to_csv(OUTPUT_PATH, index=False)
```
